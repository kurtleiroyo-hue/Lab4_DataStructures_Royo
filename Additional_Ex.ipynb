```python
import random

# ==========================================
# STUDENT-SPECIFIC INPUTS & SETUP
# ==========================================
LAST_NAME = "Royo"
STUDENT_ID = "TUPM-26-1155"
FAVORITE_ARTIST = "Adie"

# Rules implementation
SEED_NUM = int(STUDENT_ID[-1])  # 5
ID_SUM = sum(int(c) for c in STUDENT_ID if c.isdigit())  # 20
NAME_LENGTH = len(LAST_NAME)  # 4
ARTIST_LENGTH = len(FAVORITE_ARTIST)  # 4
ARTIST_SUM = sum(ord(c.upper()) - ord('A') + 1 for c in FAVORITE_ARTIST if c.isalpha())  # 23


# ==========================================
# EXERCISE 1: Engineering Test Data Organizer
# ==========================================
def exercise_1():
    print("==================================================")
    print("EXERCISE 1: Engineering Test Data Organizer")
    print("==================================================\n")

    sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
    categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

    artist_char_positions = [ord(c.upper()) - ord('A') + 1 for c in FAVORITE_ARTIST]
    random.seed(SEED_NUM)

    records = []
    for src in sources:
        for cat in categories:
            val = round(SEED_NUM * 10 + sum(artist_char_positions) + random.uniform(5, 50), 2)
            records.append({"Source": src, "Category": cat, "Value": val})

    # Include repeated record & extra measurements
    records.append(records[0].copy())
    records.append({"Source": sources[0], "Category": categories[0], "Value": round(records[0]["Value"] + 2.5, 2)})
    records.append({"Source": sources[1], "Category": categories[1], "Value": round(records[1]["Value"] + 1.2, 2)})

    print("1. Generated Test Data:")
    for r in records:
        print(r)

    # Filter duplicates
    seen = set()
    distinct_records = []
    repeated_records = []
    for r in records:
        item = (r["Source"], r["Category"], r["Value"])
        if item in seen:
            repeated_records.append(r)
        else:
            seen.add(item)
            distinct_records.append(r)

    print("\n2. Repeated Records:")
    for r in repeated_records:
        print(r)

    print("\n3. Category Summaries:")
    cat_summary = {}
    for r in distinct_records:
        cat = r["Category"]
        cat_summary.setdefault(cat, []).append(r["Value"])

    for cat, vals in cat_summary.items():
        avg = sum(vals) / len(vals)
        print(f"Category: {cat:<12} | Count: {len(vals)} | Min: {min(vals):.2f} | Max: {max(vals):.2f} | Mean: {avg:.2f}")

    print("\n" + "="*50 + "\n")
```
